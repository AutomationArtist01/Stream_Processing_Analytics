# Experiment 4: Packet Arrival Stream Processing

## Aim

To generate a stream of network packets of random sizes, process each packet as it arrives,
and visualize the packet arrival stream using a line plot.

## Theory

In a computer network, data is transmitted in small units called packets. Packets do not arrive
all at once, they arrive one after another as a continuous stream. The size of each packet is not
fixed and normally varies between 64 bytes and 1500 bytes, which is the maximum transmission unit
of an Ethernet frame. Processing such packets as they arrive is an example of stream processing.
Plotting the packet sizes in the order of arrival helps in identifying traffic patterns and bursts
in the network.

## Requirements

- Python 3.x
- Jupyter Notebook
- Random module
- Time module
- Matplotlib

```python
import random
import time
import matplotlib.pyplot as plt
```

```python
packet_stream = []

for i in range(20):

    packet_size = random.randint(64, 1500)
    packet_stream.append(packet_size)

    print(f"Packet {i + 1} arrived: {packet_size} bytes")

    time.sleep(0.2)
```

```python
plt.plot(packet_stream, marker="o")

plt.title("Network Packet Arrival Stream")
plt.xlabel("Packet Number")
plt.ylabel("Packet Size (Bytes)")
plt.grid(True)

plt.show()
```

## Interpretation

- Each packet arrives one after another.
- Packet size varies from 64 to 1500 bytes.
- Simulates real network traffic.
- Displays packet size over time.
- Helps identify traffic patterns and bursts.

## Result

The network packet arrival stream was successfully simulated and visualized. Twenty packets of
random sizes were generated and plotted, showing how packet size varies over time in a network.
