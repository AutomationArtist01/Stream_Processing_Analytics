# Experiment 9: Real-Time Bitcoin Price Analysis using a Sliding Window

## Problem Statement

A financial analyst continuously monitors the live Bitcoin price using the CoinGecko API, which
provides real-time cryptocurrency market data. Write a Python program to fetch the latest price at
regular intervals, store the most recent values in a fixed size sliding window, and perform
statistical analysis for each window to analyse short-time market fluctuations.

## Theory

Cryptocurrency prices change continuously, so the price feed behaves like a data stream. Storing
every price ever received is not practical, therefore only the most recent values are kept. A
`deque` with a `maxlen` is used for this purpose. When the deque is full and a new price is appended,
the oldest price is removed automatically, which gives a sliding window without any extra code.
Statistics such as mean, median, minimum, maximum and standard deviation are then calculated on this
window to understand the short-time behaviour of the market. The API call is wrapped in a
try-except block, because a network request can fail or the API can return an unexpected response.

## Requirements

- Python 3.x
- Jupyter Notebook
- Requests
- Time module
- Collections module
- Statistics module
- An active internet connection

```python
import requests
import time
import statistics

from collections import deque
```

```python
def get_bitcoin_price():

    url = "https://api.coingecko.com/api/v3/simple/price"

    parameters = {
        "ids": "bitcoin",
        "vs_currencies": "usd"
    }

    try:
        response = requests.get(url, params=parameters, timeout=10)
        response.raise_for_status()

        data = response.json()

        if "bitcoin" in data and "usd" in data["bitcoin"]:
            return data["bitcoin"]["usd"]
        else:
            print("Unexpected API response:", data)
            return None

    except requests.exceptions.RequestException as e:
        print("Network / API Error:", e)
        return None
```

```python
window_size = 5

price_window = deque(maxlen=window_size)

print("Window size :", window_size)
```

```python
print("Monitoring Bitcoin price...\n")
```

```python
for i in range(10):

    print(f"Reading {i + 1}")

    price = get_bitcoin_price()

    if price is not None:

        price_window.append(price)
        print("Current Price (USD):", price)

        if len(price_window) == window_size:

            print("Sliding Window :", list(price_window))
            print("Mean    :", round(statistics.mean(price_window), 2))
            print("Median  :", statistics.median(price_window))
            print("Minimum :", min(price_window))
            print("Maximum :", max(price_window))
            print("SD      :", round(statistics.stdev(price_window), 2))

    else:
        print("Price could not be fetched")

    print("-" * 40)

    time.sleep(10)
```

## Interpretation

- The CoinGecko API returns the live Bitcoin price in USD.
- The price is fetched 10 times with a gap of 10 seconds between readings.
- A `deque` with `maxlen = 5` keeps only the five most recent prices.
- When a sixth price arrives, the oldest price is dropped automatically.
- Statistics are calculated only after the window is completely filled.
- The standard deviation shows how much the price fluctuated in the recent window.
- The try-except block prevents the program from stopping when a network error occurs.

## Result

The program successfully fetched real-time Bitcoin prices using the CoinGecko API. A sliding window
of the latest five prices was maintained and updated continuously. Statistical measures such as
mean, median, minimum, maximum and standard deviation were calculated for each window. The results
displayed the current Bitcoin price along with updated statistics after every new price fetched,
enabling continuous monitoring of market trends.
