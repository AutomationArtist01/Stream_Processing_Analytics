# Experiment 2: Random Stream Data Processing Pipeline

## Aim
To generate a random data stream and visualize it using line and scatter plots.

## Theory
A data stream is a continuous flow of data generated over time. In this experiment, random values are generated and visualized to understand streaming data behavior.

## Requirements
- Python 3.x
- Jupyter Notebook
- Matplotlib

```python
import random
import matplotlib.pyplot as plt
```

```python
stream_data = []
```

```python
for i in range(100):
    value = random.randint(1, 100)
    stream_data.append(value)
```

```python
plt.plot(stream_data)
plt.title("Random Stream Data")
plt.xlabel("Data Point Number")
plt.ylabel("Random Value")
plt.grid(True)
plt.show()
```

```python
x = []
for i in range(100):
    x.append(i + 1)
```

```python
plt.scatter(x, stream_data)
plt.title("2D View of Stream Data")
plt.xlabel("Data Point Number")
plt.ylabel("Random Value")
plt.grid(True)
plt.show()
```

## Interpretation

- A stream of 100 random values is generated.
- Each value ranges from 1 to 100.
- The line plot shows how values change over time.
- The scatter plot shows the distribution of stream records.
- This approach is useful for monitoring continuously incoming data.

## Result

The random stream data was successfully generated and visualized using line and scatter plots.
