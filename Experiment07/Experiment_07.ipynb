# Experiment 7: Case Study on Incremental Learning

## Case Study

Modern organizations generate massive amounts of network traffic every second. Traditional machine
learning models used in Intrusion Detection Systems (IDS) are trained on historical data and require
complete retraining whenever new data arrives. This retraining process is time consuming,
computationally expensive and unsuitable for a real-time cyber security environment.

To address this challenge, an incremental learning approach using Stochastic Gradient Descent (SGD)
was implemented. The model continuously learns from incoming network traffic and updates itself
whenever new data arrives, without retraining on the entire dataset.

## Requirements

- Python 3.x
- Jupyter Notebook
- NumPy
- Scikit-learn

```python
import numpy as np

from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score
```

```python
np.random.seed(42)

X_train = np.random.randn(100, 2)
y_train = (X_train[:, 0] + X_train[:, 1] > 0).astype(int)

print("Training samples :", X_train.shape[0])
print("Features per sample :", X_train.shape[1])
```

**Interpretation**

- `X_train` creates 100 training samples with 2 numerical features each.
- `y_train` generates binary class labels.

```python
model = SGDClassifier(loss="log_loss")
```

**Interpretation**

- Creates an SGD classifier for binary classification.
- The `log_loss` loss makes the model perform logistic regression using SGD optimization.

```python
model.partial_fit(
    X_train,
    y_train,
    classes=np.array([0, 1])
)
```

**Interpretation**

- `partial_fit()` trains the SGD model on the initial training data and enables incremental
  learning by recognising classes 0 and 1.

```python
for i in range(20):

    X_new = np.random.randn(1, 2)
    y_new = (X_new[:, 0] + X_new[:, 1] > 0).astype(int)

    prediction = model.predict(X_new)

    print(f"Record {i + 1}:")
    print("Prediction:", prediction[0])
    print("Actual:", y_new[0])

    model.partial_fit(X_new, y_new)
```

**Interpretation**

- Simulates 20 new incoming data records.
- Predicts their class labels.
- After each prediction the model is updated incrementally using `partial_fit()` so that it learns
  from the new record.

```python
X_test = np.random.randn(50, 2)
y_test = (X_test[:, 0] + X_test[:, 1] > 0).astype(int)

prediction = model.predict(X_test)

acc = accuracy_score(y_test, prediction)

print("Accuracy:", acc)
```

**Interpretation**

- The test accuracy is approximately 0.98, that is 98%.

## Result

The incremental intrusion detection model was successfully implemented using SGD. The model was
trained on the initial data, updated itself after every new incoming record using `partial_fit()`,
and achieved a test accuracy of approximately 0.98, that is 98%, confirming that incremental
learning is suitable for real-time network traffic.
