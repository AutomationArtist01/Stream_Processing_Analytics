# Experiment 8: Statistical Analysis of Data using Sliding Window

## Aim

To implement the sliding window algorithm that performs statistical analysis of a continuously
arriving data stream, calculating the mean, median, variance, standard deviation, maximum, minimum
and range for every window.

## Statistical Analysis

The following measures are calculated for each window:

- Mean
- Median
- Variance
- Standard deviation
- Maximum
- Minimum
- Range

## Theory

In stream processing the data never ends, so statistics cannot be calculated on the complete
dataset. Instead a window of fixed size is placed over the most recent elements of the stream. As
each new element arrives the window moves forward by one position, dropping the oldest element and
including the newest one. Statistics are then recalculated for the current window only. This gives
an up to date picture of the recent behaviour of the stream while using a constant amount of memory.

## Algorithm

1. Read the streaming data.
2. Fix the window size.
3. Move the window as new data enters.
4. Extract the current window.
5. Perform statistical analysis.
6. Display the result.
7. Repeat the steps.

## Total Number of Windows Created

```text
Total windows = n - w + 1

where n = 20 and w = 5

              = 20 - 5 + 1
              = 16
```

## Requirements

- Python 3.x
- Jupyter Notebook
- Statistics module
- Random module

````python
import statistics
import random
````

````python
stream = [random.randint(10, 100) for i in range(20)]

print("Stream :", stream)
print("Total elements :", len(stream))
````

````python
window_size = 5

print("Window size :", window_size)
print("Total windows :", len(stream) - window_size + 1)
````

````python
for i in range(len(stream) - window_size + 1):

    window = stream[i : i + window_size]

    mean = statistics.mean(window)
    median = statistics.median(window)
    variance = statistics.variance(window)
    stdev = statistics.stdev(window)
    maximum = max(window)
    minimum = min(window)
    data_range = maximum - minimum

    print("\nWindow", i + 1, ":", window)
    print("Mean     :", mean)
    print("Median   :", median)
    print("Variance :", variance)
    print("Std Dev  :", stdev)
    print("Max      :", maximum)
    print("Min      :", minimum)
    print("Range    :", data_range)
````

## Interpretation

- The stream contains 20 randomly generated values between 10 and 100.
- The window size is fixed at 5.
- The window moves forward by one position as new data enters.
- Only the current window is used for the calculation, so memory usage stays constant.
- Statistics are recalculated for every window.
- A total of 16 windows are created, since 20 - 5 + 1 = 16.
- The changing mean and range across the windows show how the stream behaves over time.

## Result

The streaming data was successfully processed using the sliding window technique. Statistical
measures such as mean, median, variance, standard deviation, maximum, minimum and range were
calculated for each window.
