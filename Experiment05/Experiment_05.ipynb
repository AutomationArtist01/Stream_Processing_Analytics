# Experiment 5: Count Number of Occurrences in a Data Stream

## Aim

To process a stream of elements one by one, maintain a running count of every element using a
dictionary, and display the final frequency of each element without scanning the stream a second
time.

## Theory

In stream processing, data arrives continuously and each element can normally be read only once.
Because of this, algorithms must complete their work in a single pass over the data. Counting the
number of occurrences is one of the most basic stream operations. A dictionary is used to store the
count, where the key is the element and the value is its frequency. When an element arrives, if it
is already present in the dictionary its count is increased by one, otherwise it is inserted with a
count of one. The count is therefore always up to date and no second scan of the stream is required.

## Requirements

- Python 3.x
- Jupyter Notebook

```python
stream = ['A', 'B', 'C', 'D', 'A', 'B', 'C', 'D', 'E', 'A']

print("Stream :", stream)
print("Total elements :", len(stream))
```

```python
count = {}

print("Initial counter :", count)
```

```python
for item in stream:

    if item in count:
        count[item] += 1
    else:
        count[item] = 1

    print("Current counts:", count)
```

```python
print("\nFinal Counts")

for key, value in count.items():
    print(key, ":", value)
```

## Interpretation

- Stream elements arrive sequentially.
- Each element represents incoming data.
- A dictionary stores the occurrence count.
- Key = element.
- Value = frequency.
- Each item is processed once.
- The count updates immediately.
- No second scan is required.
- It shows the total occurrences of each element.
- A final frequency summary is generated.

## Result

The number of occurrences of each element in the data stream was successfully counted using single
pass processing. The final frequency summary was generated as A = 3, B = 2, C = 2, D = 2 and E = 1.
