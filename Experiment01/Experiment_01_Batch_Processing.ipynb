## Batch Processing Case Study – Student Marks Analysis

**Aim:** Implement batch processing using Pandas to analyze student marks.

```python
import pandas as pd
```

```python
df = pd.read_csv("students_marks.csv")
print(df.head())
```

```python
def calculate_grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 80:
        return "B"
    elif mark >= 70:
        return "C"
    elif mark >= 60:
        return "D"
    else:
        return "F"
```

```python
df["Grade"]=df["Marks"].apply(calculate_grade)
df["Status"]=df["Marks"].apply(lambda x: "Pass" if x>=40 else "Fail")
print(df.head())
```

```python
df.to_csv("processed_data.csv",index=False)
```

```python
print("Average Marks:",df["Marks"].mean())
print("Highest Marks:",df["Marks"].max())
print("Lowest Marks:",df["Marks"].min())
```

### Result
The dataset was processed in batch mode, grades and pass/fail status were generated, and summary statistics were calculated.
